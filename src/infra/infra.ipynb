```python
import os
import torch

def run_multi_depth_code_completion(infra_engine, tokenizer, target_py_path: str):
    print("\n" + "="*50)
    print("🚀 开始进行『超长代码补全』多深度长程测试")
    print("="*50)

    if not os.path.exists(target_py_path):
        print(f"❌ 找不到文件: {target_py_path}")
        return
        
    with open(target_py_path, "r", encoding="utf-8") as f:
        raw_code = f.read()

    magic_func_name = "compute_quantum_mark42_loss_v1"
    needle_definition = f"""
def {magic_func_name}(hidden_states, attention_mask):
    # This is a highly classified algorithm for context compression
    compressed_state = hidden_states * 0.99
    return compressed_state
"""
    query_trigger = f"\n# Apply the highly classified algorithm\nfinal_loss = {magic_func_name[:8]}"

    lines = raw_code.split('\n')
    total_lines = len(lines)
    
    depths = [0.0, 0.25, 0.50, 0.75, 0.90, 0.99]
    depths = [i / 100 for i in range(50)]
    results = []

    for depth in depths:
        insert_idx = int(total_lines * depth)
        
        front_code = "\n".join(lines[:insert_idx])
        middle_code = "\n".join(lines[insert_idx:])
        
        full_prompt = front_code + "\n" + needle_definition + "\n" + middle_code + query_trigger
        
        inputs_dict = tokenizer(full_prompt, add_special_tokens=False)
        input_ids_list = inputs_dict["input_ids"]
        
        print(f"\n📍 深度: {depth * 100:>5.1f}% (插入于第 {insert_idx} 行)...")
        
        try:
            generated_text = infra_engine.generate(
                input_ids=input_ids_list, 
                max_new_tokens=15, 
                repetition_penalty=1.0,
            )
            
            clean_output = generated_text.strip().split('\n')[0].split('(')[0]
            is_success = "quantum" in clean_output.lower() or "mark42" in clean_output.lower()
            
            print(f"   [模型输出] {clean_output}")
            print(f"   [判定结果] {'✅ 成功 (Hit)' if is_success else '❌ 失败 (Miss)'}")
            
            results.append((depth, is_success, clean_output))
            
        except Exception as e:
            print(f"   ⚠️ 生成出错: {e}")

    print("\n" + "="*50)
    print("📊 代码补全多深度 - 测试结果总结")
    print("="*50)
    for depth, is_success, output in results:
        status = "✅ Hit" if is_success else "❌ Miss"
        short_output = output[:40] + "..." if len(output) > 40 else output
        print(f"{depth * 100:>13.1f}% | {status:<10} | {short_output}")
    print("="*50)

# 运行测试
my_long_code_file = "/root/miniconda3/lib/python3.10/site-packages/transformers/models/llama/modeling_llama.py"
run_multi_depth_code_completion(infra_engine, tokenizer, my_long_code_file)
```

```python
import textwrap
def print_boxed(text, width=80):
    wrapper = textwrap.TextWrapper(width=width, break_long_words=False, replace_whitespace=False)
    
    paragraphs = text.split('\n')
    
    print("┏" + "━" * (width + 2) + "┓")
    for para in paragraphs:
        if not para.strip():
            print("┃ " + " " * width + " ┃")
            continue
        
        lines = wrapper.wrap(para)
        for line in lines:
            print(f"┃ {line.ljust(width)} ┃")
            
    print("┗" + "━" * (width + 2) + "┛")
```

```python
from __future__ import annotations

import argparse
import sys
from pathlib import Path
from types import SimpleNamespace

import torch


def _parse_layers(spec: str) -> list[int]:
    parts = [p.strip() for p in str(spec).split(",")]
    return [int(p) for p in parts if p]


def run():
    ap = argparse.ArgumentParser()
    ap.add_argument("--resume_path", type=str, required=True)
    ap.add_argument("--data_jsonl", type=str, default="data/wikitext_long10.jsonl")
    ap.add_argument("--out_dir", type=str, default="data/needle_scale")
    ap.add_argument("--chunk_size", type=int, default=16)
    ap.add_argument("--buffer_num", type=int, default=1)
    ap.add_argument("--q_num", type=int, default=None)
    ap.add_argument("--target_layers", type=str, default="15,23,31")
    ap.add_argument("--num_positions", type=int, default=20)
    ap.add_argument("--seed", type=int, default=0)
    ap.add_argument("--max_texts", type=int, default=None)
    ap.add_argument("--max_new_tokens", type=int, default=20)
    ap.add_argument("--temperature", type=float, default=0.0)
    ap.add_argument("--repetition_penalty", type=float, default=1.0)
    ap.add_argument("--tag", type=str, default="needle")
    args = ap.parse_args()
    print(args.tag)
    # exit()
    root = Path(__file__).resolve().parents[2]
    if str(root) not in sys.path:
        sys.path.insert(0, str(root))

    from src.experiment.needle import NeedleScaleConfig, run_needle_scale_from_jsonl
    from src.infra.infra import Mark42StreamingEngine
    from src.train_utils import load_model, load_tokenizer

    device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
    dummy_args = SimpleNamespace(
        resume_path=str(args.resume_path),
        model_name="",
        load_weights_only=True,
        phase="phase1",
        javis_num_queries=(int(args.q_num) if args.q_num is not None else 2),
        javis_target_layers=_parse_layers(args.target_layers),
        chunk_size=int(args.chunk_size),
        buffer_num=int(args.buffer_num),
    )

    tokenizer, _ = load_tokenizer(dummy_args)
    model = load_model(dummy_args, tokenizer, device, is_resume=True)
    model = model.to(torch.bfloat16)

    engine = Mark42StreamingEngine(
        model,
        tokenizer,
        chunk_size=int(args.chunk_size),
        buffer_num=int(args.buffer_num),
        q_num=(int(args.q_num) if args.q_num is not None else None),
        target_layers=tuple(_parse_layers(args.target_layers)),
    )

    cfg = NeedleScaleConfig(
        num_positions=int(args.num_positions),
        seed=int(args.seed),
        max_new_tokens=int(args.max_new_tokens),
        temperature=float(args.temperature),
        repetition_penalty=float(args.repetition_penalty),
        tag=args.tag,
    )
    meta = run_needle_scale_from_jsonl(
        engine,
        tokenizer,
        data_jsonl_path=str(args.data_jsonl),
        out_dir=str(args.out_dir),
        config=cfg,
        max_texts=(int(args.max_texts) if args.max_texts is not None else None),
    )
    print(meta)


if __name__ == "__main__":
    main()

```
