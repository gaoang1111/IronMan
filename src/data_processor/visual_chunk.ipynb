```python
def visualize_chunks(tokenizer, text, chunk_lens):
    input_ids = tokenizer.encode(text, add_special_tokens=True)

    print("=" * 80)
    print("Original text:\n")
    print(text)
    print("=" * 80)

    cur = 0
    for i, ln in enumerate(chunk_lens):
        chunk_ids = input_ids[cur:cur + ln]
        chunk_text = tokenizer.decode(chunk_ids, skip_special_tokens=False)

        print(f"\n[Chunk {i}] len={ln}")
        print("-" * 40)
        print(chunk_text)
        print("-" * 40)

        cur += ln

    print("\nTotal tokens:", len(input_ids))
    print("Sum chunk_lens:", sum(chunk_lens))
```

```python
from transformers import AutoTokenizer

tokenizer = AutoTokenizer.from_pretrained("/default-vepfs/public/user/ga/Iron/models/Llama-3.1-8B")

```

```python
import json

file_path = "/default-vepfs/public/user/ga/Iron/IronMan/data/eval_adaptive.jsonl"
target_line = 2  # 假设取第三行
current_line = 0

with open(file_path, 'r', encoding='utf-8') as f:
    for line in f:
        if current_line == target_line:
            data = json.loads(line)
            text = data.get('text')
            chunk_lens = data.get('chunk_lens')
            # print(f"Text: {text}\nChunk_lens: {chunk_lens}")
            break
        current_line += 1

# text = "your text here..."
# chunk_lens = [13, 14, 5, 7, ...]

visualize_chunks(tokenizer, text, chunk_lens)
```
